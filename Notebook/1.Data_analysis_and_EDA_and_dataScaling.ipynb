```python
from sklearn.model_selection import train_test_split


x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    # stratify=y
)
```

```python
num_features = [
    'ApplicantIncome',
    'LoanAmount',
    'Large_loan'
]
```

```python
binary_features = [
    'Gender',
    'Married',
    'Education',
    'Self_Employed',

]
ordinal_features = [
    
    'Dependency',
    
    'Income_category'
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    OrdinalEncoder
)
```

```python
num_transformer=StandardScaler()
cat_transformer=OneHotEncoder(
    handle_unknown='ignore',
    drop='first'
)
ordinal_transformer = OrdinalEncoder(
    
)
```

```python
processor=ColumnTransformer(transformers=[
    ('num', num_transformer,num_features),
    ('cat',cat_transformer,binary_features),
    ('ord',ordinal_transformer, ordinal_features)


])
```

```python
df.head()
```

```python
x_train_transformed=processor.fit_transform(x_train)
x_test_transformed=processor.transform(x_test)
```

```python
import matplotlib.pyplot as plt
import numpy as np

df['Loan_Status_encoded'] = df['Loan_Status'].map({
    'Y':1,
    'N':0
})

from sklearn.linear_model import LinearRegression

# Features and target
X = df[['ApplicantIncome']]
y = df['Loan_Status_encoded']


# Train model
model = LinearRegression()
model.fit(X, y)


# Create points for line
x_range = np.linspace(
    X['ApplicantIncome'].min(),
    X['ApplicantIncome'].max(),
    100
).reshape(-1,1)


# Predictions
y_line = model.predict(x_range)


# Plot
plt.figure(figsize=(10,6))

# Actual data points
plt.scatter(
    X,
    y,
    label="Actual Loan Status"
)

# Regression line
plt.plot(
    x_range,
    y_line,
    label="Linear Regression Line"
)


# Probability boundaries
plt.axhline(
    0,
    linestyle="--",
)

plt.axhline(
    1,
    linestyle="--",
)


plt.xlabel("Applicant Income")
plt.ylabel("Loan Status")

plt.title(
    "Applicant Income vs Loan Approval (Linear Regression)"
)

plt.legend()
plt.show()
```
